# 🖥️🧬 MÁQUINAS VIRTUALES EN BIOINFORMÁTICA

# 1️⃣ ¿Qué es una máquina virtual? 🖥️

Una **máquina virtual (Virtual Machine, VM)** es un computador simulado mediante software que funciona dentro de otro computador.

Podemos imaginarla como:

💻 **Computador físico**
⬇️
⚙️ **Sistema operativo anfitrión**
⬇️
🧩 **Software de virtualización**
⬇️
🖥️ **Máquina virtual**
⬇️
🐧 **Sistema operativo invitado**
⬇️
🧬 **Programas bioinformáticos**

Por ejemplo:

> 💻 Windows → 🖥️ VirtualBox → 🐧 Ubuntu → 🧬 BLAST

La máquina virtual permite ejecutar un sistema operativo diferente sin instalarlo directamente como sistema principal del computador.

---

# 2️⃣ ¿Cómo funciona una máquina virtual? ⚙️

Una VM utiliza los recursos del computador físico:

🧠 CPU
💾 RAM
💽 Almacenamiento
🌐 Red
🎮 GPU, cuando está disponible

El software denominado **hipervisor** administra estos recursos.

### 🔄 Esquema

```text
              💻 COMPUTADOR FÍSICO
                     │
          ┌──────────┴──────────┐
          │                     │
     🪟 Windows              🐧 Linux
          │                     │
          └──────────┬──────────┘
                     │
              ⚙️ HIPERVISOR
                     │
              🖥️ MÁQUINA VIRTUAL
                     │
                🐧 Ubuntu
                     │
          ┌──────────┼──────────┐
          │          │          │
       🧬 BLAST    🧬 BWA     🧬 SAMtools
```

💻 Hardware

* VMware ESXi

### 🥈 Tipo 2 — Hosted

Se ejecuta sobre un sistema operativo convencional.

💻 Hardware

⬇️
⚙️ VirtualBox / VMware Workstation
⬇️

#📦 ¿Qué es Conda?
Conda es un gestor de paquetes, dependencias y entornos que permite instalar, actualizar y organizar programas y librerías, especialmente útil en Bioinformática.

En términos sencillos:

🧰 Conda es como un administrador de herramientas para crear ambientes de trabajo separados y controlar qué versiones de los programas utiliza cada proyecto.

Diferentes programas pueden **requerir versiones diferentes de una misma dependencia**. Esto puede generar conflictos.

Conda permite crear ambientes independientes:


# 1️⃣7️⃣ ⚠️ Limitaciones de las máquinas virtuales

Las VMs son muy útiles, pero no son perfectas.

### 🧠 1. Consumo de RAM

La VM necesita memoria.

Si tenemos:

💻 8 GB RAM

y asignamos:

🖥️ VM = 6 GB

el sistema anfitrión tendrá menos memoria disponible.

---

### ⚙️ 2. Rendimiento

Existe cierta sobrecarga debido a la virtualización.

Por eso:

🖥️ VM ≠ rendimiento idéntico al hardware físico.

---

### 💽 3. Almacenamiento

Una VM puede necesitar:

📦 20 GB
📦 50 GB
📦 100 GB o más

Los datos NGS pueden aumentar rápidamente el espacio requerido.

---

### 🌐 4. Conectividad

En la nube:

☁️ VM + Internet

La transferencia de grandes archivos puede ser lenta o costosa.

---

### 💰 5. Costos

Las máquinas virtuales locales pueden ser gratuitas, pero las VMs en la nube normalmente tienen costos asociados.

⚠️ Más:

CPU

* RAM
* almacenamiento
* transferencia
  = 💰 mayor costo

---

### 🔐 6. Seguridad

Los datos biológicos pueden contener información sensible.

Por ello debemos considerar:

🔐 autenticación
🔐 permisos
🔐 cifrado
🔐 copias de seguridad
🔐 protección de datos

### 🖥️  ¿Qué es una API?

Una **API (Application Programming Interface)**, o **Interfaz de Programación de Aplicaciones**, es un conjunto de reglas y mecanismos que permite que **dos programas o sistemas se comuniquen entre sí**.

En términos sencillos:

> **Una API permite que un programa solicite información o una acción a otro programa sin que el usuario tenga que hacerlo manualmente.**

## Ejemplo sencillo

Imagina un restaurante:

* Tú = usuario
* Cocina = servidor
* Menú = API
* Mesero = intermediario

Tú no entras directamente a la cocina. Consultas el menú y haces un pedido al mesero. El mesero lleva la solicitud a la cocina y luego te devuelve el resultado.

De manera similar:

```text
Programa A
    ↓
   API
    ↓
Servidor / Programa B
    ↓
Respuesta
    ↓
Programa A
```

## 🧬 Ejemplo en Bioinformática

Supongamos que queremos obtener información sobre genes desde una base de datos en Internet.

En lugar de entrar manualmente al sitio web y buscar cada gen, podemos utilizar una API:

```text
Python
   ↓
API
   ↓
Base de datos biológica
   ↓
Información del gen
   ↓
Python
```

Por ejemplo, un programa podría solicitar:

```text
"Obtener información del gen BRCA1"
```

La API recibe la solicitud y devuelve información que nuestro programa puede procesar.

### 🧬 ¿Para qué sirven las API en Bioinformática?

Son muy importantes porque permiten automatizar el acceso a:

* 🧬 Bases de datos genómicas
* 🧪 Bases de datos de proteínas
* 🌱 Datos de biodiversidad
* 🧬 Secuencias de ADN y ARN
* 📊 Datos científicos
* 🔬 Servicios de análisis
* ☁️ Plataformas de computación

Por ejemplo, una API puede permitir que un programa consulte automáticamente miles de registros en lugar de que un investigador los busque uno por uno.

### API vs programa

| Concepto      | Función                                          |
| ------------- | ------------------------------------------------ |
| Programa      | Realiza una tarea                                |
| API           | Permite que otros programas se comuniquen con él |
| Servidor      | Procesa solicitudes y proporciona recursos       |
| Base de datos | Almacena información                             |
| Usuario       | Solicita o utiliza los resultados                |

### 🔑 Para recordar

**API = un puente de comunicación entre programas.**

En Bioinformática:

```text
🐍 Python/R/bash u otro.
   ↓
🔌 API
   ↓
🧬 Base de datos
   ↓
📊 Datos biológicos
   ↓
🐍 Python
```

Una API es especialmente útil cuando queremos **automatizar la consulta y descarga de grandes cantidades de información biológica**.
