# 🔬 17. Ejemplo: análisis de secuencias

### Pregunta científica

> ¿Cuál de varias secuencias tiene mayor contenido GC?

### Entrada 📥
Varias secuencias de ADN.

### Procesamiento ⚙️
- Contar G y C.
- Calcular longitud.
- Calcular GC%.

### Salida 📤
Una tabla o informe con los resultados.


`````python
secuencias = [
    "ATGCGATCGATCG",
    "ATATATATATAT",
    "GCGCGCGCGCGC",
    "ATGCTAGCTAGCTAGC"
]

for i, secuencia in enumerate(secuencias, start=1):
    gc = (secuencia.count("G") + secuencia.count("C")) / len(secuencia) * 100
    print(f"🧬 Secuencia {i}: {len(secuencia)} pb | 📊 GC: {gc:.2f}%")
`````

# 🏆  Analizador de ADN

Construye un programa que analice:

```python
secuencias = [
    "ATGCGATCGATCG",
    "ATATATATATAT",
    "GCGCGCGCGCGC",
    "ATGCTAGCTAGCTAGC"
]
```

El programa debe:

1. 🔁 Recorrer todas las secuencias.
2. 📏 Calcular longitud.
3. 🅰️ Contar A.
4. 🆃 Contar T.
5. 🅶 Contar G.
6. 🅲 Contar C.
7. 📊 Calcular GC%.
8. 🔀 Clasificar GC.
9. 📋 Mostrar resultados organizadamente.


`````python
secuencias = [
    "ATGCGATCGATCG",
    "ATATATATATAT",
    "GCGCGCGCGCGC",
    "ATGCTAGCTAGCTAGC"
]

# 🔁 Recorrer todas las secuencias
for i, secuencia in enumerate(secuencias, start=1):

    # 📏 Longitud
    longitud = len(secuencia)

    # 🅰️🆃🅶🅲 Conteo de nucleótidos
    A = secuencia.count("A")
    T = secuencia.count("T")
    G = secuencia.count("G")
    C = secuencia.count("C")

    # 📊 Contenido GC
    GC = ((G + C) / longitud) * 100

    # 🔀 Clasificación
    if GC < 40:
        clasificacion = "🔵 GC bajo"
    elif GC <= 60:
        clasificacion = "🟢 GC medio"
    else:
        clasificacion = "🔴 GC alto"

    # 📋 Resultados
    print(f"\n🧬 Secuencia {i}")
    print(f"📋 Secuencia: {secuencia}")
    print(f"📏 Longitud: {longitud} pb")
    print(f"🅰️ A = {A}")
    print(f"🆃 T = {T}")
    print(f"🅶 G = {G}")
    print(f"🅲 C = {C}")
    print(f"📊 GC% = {GC:.2f}%")
    print(f"🔀 Clasificación = {clasificacion}")
`````

`````python
secuencia = "atgcgatcgatc"

print("Original:", secuencia)

# 🔠 Mayúsculas
print("upper():", secuencia.upper())

# 🔡 Minúsculas
print("lower():", secuencia.lower())

# 🅰️ Contar A
print("count('A'):", secuencia.upper().count("A"))

# 🔍 Buscar una secuencia
print("find('G'):", secuencia.upper().find("G"))

# 🔄 Reemplazar
print("replace():", secuencia.upper().replace("T", "U"))

# 🔎 Comprobar inicio
print("startswith('AT'):", secuencia.upper().startswith("AT"))

# 🔎 Comprobar final
print("endswith('TC'):", secuencia.upper().endswith("TC"))
`````
