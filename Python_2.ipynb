## 1. 📋 Métodos frecuentes de listas

Una **lista** (`list`) permite almacenar varios elementos en un orden determinado.

| Método | Función | Ejemplo |
|---|---|---|
| `append()` | Agrega un elemento al final | `lista.append("A")` |
| `extend()` | Agrega varios elementos | `lista.extend(["C","G"])` |
| `insert()` | Inserta en una posición | `lista.insert(1, "T")` |
| `remove()` | Elimina un valor | `lista.remove("A")` |
| `pop()` | Elimina y devuelve un elemento | `lista.pop()` |
| `sort()` | Ordena la lista | `lista.sort()` |
| `reverse()` | Invierte el orden | `lista.reverse()` |
| `index()` | Busca la posición de un valor | `lista.index("G")` |
| `count()` | Cuenta ocurrencias | `lista.count("A")` |
| `copy()` | Crea una copia | `copia = lista.copy()` |
| `clear()` | Vacía la lista | `lista.clear()` |

### 🧬 Ejemplo


````python
secuencias = ["ATG", "GCA", "TTA"]

secuencias.append("CCC")
secuencias.insert(1, "AAA")

print(secuencias)
print("Número de elementos:", len(secuencias))
````

## 2. 🔤 Métodos frecuentes de cadenas (`str`)

Las cadenas son fundamentales para trabajar con secuencias de ADN, ARN y proteínas.

| Método | Función |
|---|---|
| `upper()` | Convierte a mayúsculas |
| `lower()` | Convierte a minúsculas |
| `replace()` | Reemplaza texto |
| `split()` | Divide una cadena |
| `join()` | Une elementos |
| `strip()` | Elimina espacios al inicio/final |
| `find()` | Busca una subcadena |
| `count()` | Cuenta ocurrencias |
| `startswith()` | Comprueba el inicio |
| `endswith()` | Comprueba el final |

### 🧬 Ejemplo con una secuencia de ADN


````python
dna = "atgcccggatga"

print("Original:", dna)
print("Mayúsculas:", dna.upper())
print("Número de G:", dna.upper().count("G"))
print("¿Empieza por ATG?:", dna.upper().startswith("ATG"))
print("Reemplazo:", dna.upper().replace("T", "U"))
````

## 3. 📚 Métodos frecuentes de diccionarios (`dict`)

Los diccionarios almacenan información mediante pares **clave → valor**.

| Método | Función |
|---|---|
| `keys()` | Devuelve las claves |
| `values()` | Devuelve los valores |
| `items()` | Devuelve clave y valor |
| `get()` | Obtiene un valor de forma segura |
| `update()` | Agrega o modifica elementos |
| `pop()` | Elimina una clave |
| `clear()` | Vacía el diccionario |
| `copy()` | Copia el diccionario |

### 🧬 Ejemplo


````python
gen = {
    "id": "BRCA1",
    "organismo": "Homo sapiens",
    "cromosoma": 17,
    "funcion": "reparación del ADN"
}

print(gen.keys())
print(gen.values())
print(gen.items())
````

## 4. 🔑 ¿Cómo acceder a un diccionario?

Existen varias formas.

### Acceso directo con `[]`
Se utiliza cuando conocemos la clave y esperamos que exista.

```python
gen["id"]
```

### Acceso seguro con `.get()`
Es útil cuando la clave podría no existir.

```python
gen.get("id")
gen.get("expresion", "No disponible")
```

### Acceder a varios elementos


````python
gen = {
    "id": "BRCA1",
    "organismo": "Homo sapiens",
    "cromosoma": 17,
    "funcion": "reparación del ADN"
}

print("ID:", gen["id"])
print("Organismo:", gen.get("organismo"))
print("Expresión:", gen.get("expresion", "No disponible"))

print("\nRecorriendo el diccionario:")
for clave, valor in gen.items():
    print(clave, "→", valor)
````

## 6. 🧺 Métodos frecuentes de conjuntos (`set`)

Los conjuntos almacenan elementos **sin duplicados**.

| Método | Función |
|---|---|
| `add()` | Agrega un elemento |
| `remove()` | Elimina un elemento |
| `discard()` | Elimina sin producir error si no existe |
| `union()` | Unión |
| `intersection()` | Intersección |
| `difference()` | Diferencia |

### 🧬 Ejemplo


````python
genes_a = {"BRCA1", "TP53", "EGFR"}
genes_b = {"TP53", "EGFR", "KRAS"}

print("Unión:", genes_a.union(genes_b))
print("Intersección:", genes_a.intersection(genes_b))
print("Solo en A:", genes_a.difference(genes_b))
````

# 8.📂 Trabajar con archivos en Python

Python permite trabajar con archivos de texto, CSV, FASTA, FASTQ y muchos otros formatos.

La función básica es:

open("archivo.txt", "modo")

Algunos modos frecuentes:

| Modo | Función |
|---|---|
| `"r"` | 📖 Lectura |
| `"w"` | ✍️ Escritura; sobrescribe el archivo |
| `"a"` | ➕ Agregar contenido al final |
| `"x"` | 🆕 Crear un archivo nuevo |
| `"rb"` | 💾 Lectura binaria |

En bioinformática, los archivos pueden contener secuencias, conteos de expresión, variantes o metadatos.

````python
# 📂 Crear un archivo de ejemplo

texto = """Muestra,Gen,Expresion
M01,TP53,12.5
M02,TP53,18.2
M03,TP53,25.7
"""

with open("expresion.txt", "w", encoding="utf-8") as archivo:
    archivo.write(texto)

print("✅ Archivo creado correctamente")
````

##9. 📖 Abrir y leer un archivo

Podemos utilizar `open()` y después leer su contenido con `.read()`.

````python
archivo = open("expresion.txt", "r", encoding="utf-8")

contenido = archivo.read()

print(contenido)

archivo.close()
````

### 🔒 ¿Por qué cerrar un archivo?

`close()` libera los recursos asociados al archivo. Sin embargo, existe una forma más segura y recomendada: utilizar `with`.

## 3. 🛡️ La estructura `with open()`

Cuando usamos:

```python
with open(...) as archivo:
    ...
```

Python gestiona automáticamente el cierre del archivo, incluso si ocurre un error dentro del bloque.


````python
# ✅ Forma recomendada

with open("expresion.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
````

## 10. 📖 Leer línea por línea

Para archivos grandes, leer todo el contenido simultáneamente puede no ser conveniente. Podemos procesar las líneas una por una.

Esto es especialmente importante en bioinformática porque archivos FASTA y FASTQ pueden contener millones de líneas.

````python
with open("expresion.txt", "r", encoding="utf-8") as archivo:
    for numero, linea in enumerate(archivo, start=1):
        print(numero, linea.strip())
````

## 11. ✍️ Escribir y agregar información

Con `"w"` podemos crear o sobrescribir un archivo.

⚠️ **Importante:** `"w"` elimina el contenido anterior del archivo.

Con `"a"` agregamos contenido al final.

````python
# ✍️ Agregar una nueva muestra

nueva_linea = "M04,TP53,31.4\n"

with open("expresion.txt", "a", encoding="utf-8") as archivo:
    archivo.write(nueva_linea)

print("✅ Nueva línea agregada")
````

# 🐼 8. Pandas: análisis de datos

- `Series` → una columna.

## 🔍 Seleccionar y filtrar datos

Pandas permite responder preguntas biológicas mediante filtros.

Ejemplo:

> ¿Qué muestras presentan una expresión mayor que 20?


````python
# 🔎 Filtrar por expresión

alta_expresion = df[df["Expresion"] > 20]

display(alta_expresion)
````

````python
# 🧬 Filtrar muestras de tratamiento

tratamiento = df[df["Condicion"] == "Tratamiento"]

display(tratamiento)
````

## 📊 Agrupar y resumir

Podemos calcular estadísticas por condición.

Por ejemplo, la expresión media del gen en cada grupo.

````python
resumen = (
    df.groupby("Condicion")["Expresion"]
      .agg(["count", "mean", "std", "min", "max"])
)

display(resumen)
````

# 📈 12. Matplotlib: visualización de datos

Matplotlib permite convertir datos numéricos en gráficos.

Una estructura frecuente es:

```python

plt.xlabel("...")
plt.ylabel("...")

```

La visualización permite identificar patrones que pueden ser difíciles de detectar únicamente en una tabla.

# 🧬 14. Mini-flujo bioinformático

Ahora combinaremos todo:

📂 **Archivo → 🐼 Pandas → 🔎 análisis → 📈 visualización**

Este flujo aparece constantemente en bioinformática y análisis de datos ómicos.

````python
# 🧬 Flujo completo

import pandas as pd
import matplotlib.pyplot as plt

# 1️⃣ Leer datos
datos = pd.read_csv("datos_expresion.csv")

# 2️⃣ Explorar
print("Número de muestras:", len(datos))
print("Genes:", datos["Gen"].unique())

# 3️⃣ Calcular promedio por condición
media = datos.groupby("Condicion")["Expresion"].mean()

# 4️⃣ Mostrar resultados
display(media.to_frame("Expresion_media"))

# 5️⃣ Visualizar
plt.figure(figsize=(7, 5))
plt.bar(media.index, media.values)
plt.xlabel("Condición")
plt.ylabel("Expresión media")
plt.title("Comparación de expresión")
plt.show()
````
