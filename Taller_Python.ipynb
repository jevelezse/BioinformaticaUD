# 🔬 12. Reto final — Clasificación del contenido GC

Modifica el analizador para clasificar el contenido GC:

- 🟡 GC < 40% → GC bajo
- 🟢 40–60% → GC moderado
- 🔴 GC > 60% → GC alto

````python
# ✏️ Completa el programa
def analizar_adn(Parametro):
    longitud = len(Parametro)
    A = Parametro.count("A")
    T = Parametro.count("T")
    G = Parametro.count("G")
    C = Parametro.count("C")
    GC = (G + C) / longitud * 100

    # Clasificación del contenido GC
    if GC < 40:
        clasificacion = "🟡 GC bajo"
    elif GC <= 60:
        clasificacion = "🟢 GC moderado"
    else:
        clasificacion = "🔴 GC alto"

    print("🧬 ANÁLISIS DE ADN")
    print("----------------")
    print("📏 Longitud:", longitud)
    print("🅰️ A:", A)
    print("🆃 T:", T)
    print("🅶 G:", G)
    print("🅲 C:", C)
    print("📊 GC%:", round(GC, 2))
    print("📌 Clasificación:", clasificacion)


secuencia = "ATGCGATACGCTAGCTAGCTAG"
analizar_adn(secuencia)
````

Tu notebook debe contener:

### 📊 Resultados

### 📝 Interpretación

````python
# 🎓 PRODUCTO FINAL
# 🧬 Análisis de secuencias de ADN

# Lista de secuencias
secuencias = [
    "ATGCGATCGATCG",
    "ATATATATATAT",
    "GCGCGCGCGCGC",
    "ATGCTAGCTAGCTAGC"
]


# 🧬 Función para analizar una secuencia
def analizar_adn(secuencia):

    # Cálculo de la longitud
    longitud = len(secuencia)

    # Conteo de nucleótidos
    A = secuencia.count("A")
    T = secuencia.count("T")
    G = secuencia.count("G")
    C = secuencia.count("C")

    # Cálculo del contenido GC
    GC = (G + C) / longitud * 100

    # Clasificación del contenido GC
    if GC < 40:
        clasificacion = "🟡 GC bajo"
    elif GC <= 60:
        clasificacion = "🟢 GC moderado"
    else:
        clasificacion = "🔴 GC alto"

    # Mostrar resultados
    print("🧬 Secuencia:", secuencia)
    print("📏 Longitud:", longitud, "pb")
    print("🅰️ A:", A)
    print("🆃 T:", T)
    print("🅶 G:", G)
    print("🅲 C:", C)
    print("📊 GC%:", round(GC, 2), "%")
    print("🔎 Clasificación:", clasificacion)
    print("-" * 40)


# 🔁 Analizar todas las secuencias
print("🎓 ANÁLISIS DE SECUENCIAS DE ADN")
print("=" * 40)

for secuencia in secuencias:
    analizar_adn(secuencia)


# 📝 Interpretación
print("\n📝 INTERPRETACIÓN")
print("=" * 40)
print("Las secuencias presentan diferencias en su composición de nucleótidos.")
print("La secuencia ATATATATATAT presenta un contenido GC de 0%,")
print("por lo que se clasifica como GC bajo.")
print("La secuencia GCGCGCGCGCGC presenta un contenido GC de 100%,")
print("por lo que se clasifica como GC alto.")
print("Las otras dos secuencias presentan valores intermedios,")
print("por lo que corresponden a un contenido GC moderado.")
````
