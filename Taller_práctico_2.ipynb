# 📚 Documentación del código en programación

##  ¿Qué es y por qué es importante?

La **documentación del código** consiste en comentarios e instrucciones que explican qué hace un programa, cómo utilizarlo, qué datos necesita y qué resultados genera.

En **programación y bioinformática**, permite que el código sea:

* 🧠 **Comprensible:** facilita entender cada parte.
* 🔧 **Mantenible:** permite corregir y modificar el código.
* ♻️ **Reproducible:** facilita repetir un análisis.
* 👥 **Colaborativo:** otras personas pueden comprenderlo.
* 🔄 **Reutilizable:** permite aplicarlo a nuevos datos.

En bioinformática, un análisis puede seguir:

```text
Datos → Control de calidad → Filtrado → Alineamiento → Análisis → Resultados
```

La documentación permite conocer qué se hizo en cada etapa.

---

## ¿Qué se debe documentar?

| Elemento      | ¿Qué documentar?                |
| ------------- | ------------------------------- |
| Script        | Propósito general               |
| Autor y fecha | Responsable y fecha             |
| Versión       | Versión del programa            |
| Entradas      | Archivos o parámetros           |
| Salidas       | Resultados generados            |
| Dependencias  | Programas necesarios            |
| Comandos      | Función de comandos importantes |
| Parámetros    | Significado de cada opción      |
| Ejemplo       | Forma de ejecutar el script     |

## Documentación dentro de un script 🖥️

En **Bash**, los comentarios comienzan con `#` y no se ejecutan.

```bash
# Crear el directorio para almacenar los resultados
mkdir resultados
```

El comentario debe explicar **el propósito del código**, no repetirlo.

### ❌ Poco útil

```bash
# Crear carpeta
mkdir resultados
```

### ✅ Mejor

```bash
# Crear el directorio para almacenar los archivos generados durante el análisis
mkdir resultados
```

## Encabezado del script

Un script puede comenzar con una ficha técnica:

```bash
#!/bin/bash

########################################################
# Nombre: analisis_fasta.sh
# Descripción:
#     Calcula información básica de un archivo FASTA.
#
# Autor: Nombre del estudiante
# Fecha: 24-08-2026
# Versión: 1.0
#
# Uso:
#     ./analisis_fasta.sh archivo.fasta
#
# Entrada: Archivo FASTA
# Salida: Número de secuencias
########################################################
```

También se deben indicar las herramientas necesarias:

```bash
# Requisitos:
# Bash >= 4.0
# wget
# grep
# awk
```

## Organizar y comentar el código

Dividir el script en secciones facilita su comprensión:

```bash
# ==========================================
# 1. CREACIÓN DE DIRECTORIOS
# ==========================================
mkdir -p resultados
mkdir -p datos

# ==========================================
# 2. DESCARGA DE DATOS
# ==========================================
wget -O datos/secuencias.fasta "URL"

# ==========================================
# 3. ANÁLISIS
# ==========================================
# Contar el número de secuencias
grep -c "^>" datos/secuencias.fasta

# ==========================================
# 4. RESULTADOS
# ==========================================
grep -c "^>" datos/secuencias.fasta > resultados/numero_secuencias.txt

##  Buenas prácticas ✅

| Se recomienda               | Evitar                        |
| --------------------------- | ----------------------------- |
| Usar nombres descriptivos   | Variables ambiguas            |
| Explicar entradas y salidas | Suponer conocimientos previos |
| Registrar versiones         | Omitir versiones              |
| Incluir ejemplos de uso     | No explicar cómo ejecutar     |
| Usar comentarios útiles     | Comentarios innecesarios      |
| Incluir mensajes de error   | Errores poco claros           |

📄 ¿Qué es un archivo README?

Un README es un archivo de texto que contiene información básica y explicativa sobre un proyecto, programa, conjunto de datos o repositorio.

El nombre proviene del inglés **“read me”**, que significa **“léeme”**. Su propósito es que una persona pueda entender el proyecto antes de comenzar a utilizarlo.

Los nombres más comunes son:

README.txt

README.md

README

**md** significa Markdown

💻 ¿Qué información puede contener?

Un README puede explicar:

Información	Ejemplo
| Información     | Ejemplo                             |
| --------------- | ----------------------------------- |
| 🎯 Objetivo     | ¿Para qué sirve el proyecto?        |
| 📋 Descripción  | ¿Qué contiene?                      |
| 📁 Estructura   | ¿Qué contiene cada carpeta?         |
| ⚙️ Instalación  | ¿Cómo se prepara el proyecto?       |
| ▶️ Uso          | ¿Cómo se ejecuta?                   |
| 📦 Dependencias | ¿Qué programas necesita?            |
| 🧬 Datos        | ¿De dónde proceden los datos?       |
| 📊 Resultados   | ¿Qué resultados genera?             |
| 👤 Autores      | ¿Quién desarrolló el proyecto?      |
| 📚 Referencias  | ¿Qué fuentes se utilizaron?         |
| 📄 Licencia     | ¿Cómo puede utilizarse el proyecto? |



⭐ ¿Por qué es importante en computación?

El README funciona como la “carta de presentación” de un proyecto.

1. 🧠 Facilita la comprensión

Permite que una persona nueva entienda rápidamente el proyecto.

2. 👥 Facilita el trabajo colaborativo

Si varias personas trabajan en el mismo proyecto, el README proporciona instrucciones comunes.

3. 🔄 Facilita la reproducibilidad
